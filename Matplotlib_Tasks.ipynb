```python
season_data = ["Summer", "Winter", "Automn", "Spring"]
tempreture = [49, 23, 33, 29]
plt.title("Analysing Season Data on the basis of their Temperature")
plt.xlabel('season_data')
plt.ylabel("tempreture")
plt.pie(tempreture, labels=season_data, autopct = '%1.1f%%')
plt.show()
```

```python
import matplotlib.pyplot as plt

languages = ["Python", "JavaScript", "C++", "C#", "C"]
popularity = [90, 80, 68, 62, 58]

plt.figure(figsize=(5,2))
plt.barh(languages, popularity, color=["blue", "black", "orange", "brown", "green"])

plt.title("Programming Languages Popularity")
plt.xlabel("Popularity Score")
plt.ylabel("Languages")

plt.show()
```

```python
import matplotlib.pyplot as plt
days = ["day 1", "day 2", "day 3"]
pizza_sales = [4000, 6000, 8000]
burger_sales = [3400, 4660, 9000]
plt.plot(days, pizza_sales, color="red", label="pizza sales")
plt.plot(days, burger_sales, color="yellow", label = "burgers_sales")
plt.legend()
plt.title("Restraunt Daily Sales ")
plt.xlabel("days")
plt.ylabel("items sold price")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "Ahmed", "Ayesha"]
marks = [70, 90, 75, 85]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(students, marks, color='skyblue')
plt.title("Students Marks (Bar)")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.subplot(1, 2, 2)
plt.plot(students, marks, color='red', marker ="o" )
plt.title("Students Marks (Line)")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.tight_layout()
plt.show()
```
