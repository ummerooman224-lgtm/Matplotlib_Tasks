```python
import pandas as pd
import matplotlib.pyplot as plt

Data = {
    "course": ["Python", "Ai", "Web", "Design"],
    "marks": [40, 55, 35, 25]
}

df = pd.DataFrame(Data)
display(df)

plt.bar(df["course"], df["marks"])
plt.title("Students Marks")
plt.xlabel("course")
plt.ylabel("marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["Ali", "Sara", "Ahmed", "Ayesha"]
marks = [70, 90, 75, 85]

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.bar(students, marks, color='skyblue')
plt.title("Students Marks (Bar)")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.subplot(1, 2, 2)
plt.plot(students, marks, color='red', marker ="o" )
plt.title("Students Marks (Line)")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
students = ['Maria', 'Sana', 'Muhammad', 'Dawood']
marks = [85, 92, 78, 88]
plt.plot(students, marks, color='red', marker='o')
plt.grid(True)

plt.xlabel('Students')
plt.ylabel('Marks')
plt.title('Student Marks')

plt.show()
```

```python
import matplotlib.pyplot as plt

categories = ['Food', 'Transport', 'Shopping', 'Other']
expenses = [40, 20, 25, 15]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']

plt.figure(figsize=(6, 6))
plt.pie(expenses, labels=categories, autopct='%1.1f%%', startangle=140, colors=colors)
plt.title('Expense Distribution')
plt.show()
```
